FASTopic is a fast, adaptive, stable, and transferable topic model, different
from the previous conventional (LDA), VAE-based (ProdLDA, ETM), or clustering-based (Top2Vec, BERTopic) methods.
It leverages optimal transport between the document, topic, and word embeddings from pretrained Transformers to model topics and topic distributions of documents.

Check our paper **[[NeurIPS 2024] FASTopic: Pretrained Transformer is a Fast, Adaptive, Stable, and Transferable Topic Model](https://arxiv.org/pdf/2405.17978.pdf)**

# Install spacy and download a pipeline for Chinese

We use `zh_core_web_sm` of spaCy to tokenize Chinese docs and remove stop words.

Check the pipelines for other languages in [spaCy](https://spacy.io/models).

```python
!pip install datasets
!pip install spacy
!python -m spacy download zh_core_web_sm

import spacy

zh_core = spacy.load("zh_core_web_sm")
zh_tokenizer = lambda x: [token.text for token in zh_core(x) if not token.is_stop]
```

# Download a Dataset

Download the dataset, **TNews**, which includes Chinese news.

```python
from fastopic import FASTopic
from topmost import Preprocess
from datasets import load_dataset

dataset = load_dataset("clue", "tnews", split='validation')
docs = [x['sentence'] for x in dataset]
```

```python
docs[:5]
```

# Train FASTopic

Pass the tokenizer for Chinese to preprocessing, and set the document embedding model as `paraphrase-multilingual-MiniLM-L12-v2`, which supports 50+ languages.

```python
preprocess = Preprocess(vocab_size=10000, tokenizer=zh_tokenizer)

model = FASTopic(50, preprocess, doc_embed_model='paraphrase-multilingual-MiniLM-L12-v2', verbose=True)
top_words, train_theta = model.fit_transform(docs)
```

```python
model.get_topic(topic_idx=42)
```
